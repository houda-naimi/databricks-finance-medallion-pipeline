# Gold-layer risk summary table

This notebook creates a Gold-layer risk summary table for banking risk analytics and reporting.

The purpose is to aggregate customer credit risk information from the Silver-layer credit bureau data and generate summarized KPIs by risk category.

The notebook calculates:

- total customers per risk grade
- average credit score
- total active external loans
- total overdue amount

The final table created is : banking.gold.risk_customer_summary

```python
%sql
CREATE OR REPLACE TABLE banking.gold.risk_customer_summary AS

-- — Aggregate Risk Metrics
-- This query groups customers by risk_grade and calculates risk KPIs.
-- Metrics:
-- - total_customers
-- - avg_credit_score
-- - total_external_loans
-- - total_overdue_amount
-- Source:
-- banking.silver.credit_bureau_reports

SELECT
    risk_grade,
    COUNT(customer_id) AS total_customers,
    AVG(credit_score) AS avg_credit_score,
    SUM(external_active_loans) AS total_external_loans,
    SUM(external_overdue_amount) AS total_overdue_amount
FROM banking.silver.credit_bureau_reports
GROUP BY risk_grade
```

## Validate Table Creation
